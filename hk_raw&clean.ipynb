```python
import pandas as pd
hk_price_index = pd.read_csv('Data/Raw/hk_priceindices_byclass.csv', skiprows=1)
print(hk_price_index.columns.tolist())
print(hk_price_index.head())

# Keep just quarter and All Classes
hk_price_index = hk_price_index[['Quarter', 'All Classes']].copy()
hk_price_index = hk_price_index.rename(columns={'All Classes': 'hk_price_index'})

# Convert "10-12/1979" style to "1979Q4"
def convert_hk_quarter(q):
    month_range, year = q.split('/')
    start_month = int(month_range.split('-')[0])
    quarter_num = (start_month - 1) // 3 + 1
    return f"{year}Q{quarter_num}"

hk_price_index['quarter'] = hk_price_index['Quarter'].apply(convert_hk_quarter)
hk_price_index = hk_price_index[['quarter', 'hk_price_index']]
print(hk_price_index.head())
print(hk_price_index.tail())
```

```python
hk_price_index.to_csv('Data/Raw/hongkong_price_index.csv', index=False)
```

```python
import pandas as pd

# Load with the header rows skipped (Year row, Month row, section-label row)
hk_cpi_raw = pd.read_csv('Data/Raw/hk_cpi.csv', skiprows=2, header=None)

print(hk_cpi_raw.shape)
print(hk_cpi_raw.iloc[0, :20])   # Year row
print(hk_cpi_raw.iloc[1, :20])   # Month row
print(hk_cpi_raw.iloc[2, :5])    # first data row (All items)
```

```python
year_row = hk_cpi_raw.iloc[0]
month_row = hk_cpi_raw.iloc[1]

# Find the "All items" row
all_items_row_idx = hk_cpi_raw[hk_cpi_raw[2] == 'All items'].index[0]
all_items = hk_cpi_raw.iloc[all_items_row_idx]

# Build a clean dataframe: only keep columns where Month is a real month name (not NaN)
records = []
for col in hk_cpi_raw.columns:
    month_val = month_row[col]
    year_val = year_row[col]
    if pd.notna(month_val) and month_val in ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']:
        records.append({'year': year_val, 'month': month_val, 'cpi': all_items[col]})

hk_cpi_monthly = pd.DataFrame(records)
print(hk_cpi_monthly.shape)
print(hk_cpi_monthly.head())
print(hk_cpi_monthly.tail())
```

```python
print(hk_cpi_monthly['cpi'].dtype)
print(hk_cpi_monthly[pd.to_numeric(hk_cpi_monthly['cpi'], errors='coerce').isna()])
```

```python
hk_cpi_monthly['cpi'] = pd.to_numeric(hk_cpi_monthly['cpi'])
print(hk_cpi_monthly['cpi'].dtype)  # should now show float64
```

```python
month_map = {
    'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6,
    'Jul': 7, 'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12
}

hk_cpi_monthly['month_num'] = hk_cpi_monthly['month'].map(month_map)
hk_cpi_monthly['quarter_num'] = (hk_cpi_monthly['month_num'] - 1) // 3 + 1
hk_cpi_monthly['quarter'] = hk_cpi_monthly['year'].astype(str) + 'Q' + hk_cpi_monthly['quarter_num'].astype(str)

hk_cpi = hk_cpi_monthly.groupby('quarter')['cpi'].mean().reset_index()
hk_cpi = hk_cpi.rename(columns={'cpi': 'hk_cpi'})
hk_cpi = hk_cpi.sort_values('quarter').reset_index(drop=True)

print(hk_cpi.shape)
print(hk_cpi.head())
print(hk_cpi.tail())
```

```python
hk_cpi.to_csv('Data/Raw/hongkong_cpi.csv', index=False)
```

```python
hk_gdp_raw = pd.read_csv('Data/Raw/hk_gdp.csv', skiprows=2, header=None)
print(hk_gdp_raw.shape)
print(hk_gdp_raw.iloc[0, :20])   # Year row
print(hk_gdp_raw.iloc[1, :20])   # Quarter row
print(hk_gdp_raw.iloc[2, :5])    # check where "GDP" row lands
```

```python
year_row = hk_gdp_raw.iloc[0]
quarter_row = hk_gdp_raw.iloc[1]

# Find the row where column 2 == 'GDP' (the headline series, not a component)
gdp_row_idx = hk_gdp_raw[hk_gdp_raw[2] == 'GDP'].index[0]
gdp_row = hk_gdp_raw.iloc[gdp_row_idx]

records = []
for col in hk_gdp_raw.columns:
    q_val = quarter_row[col]
    y_val = year_row[col]
    if pd.notna(q_val) and str(q_val).strip() in ['Q1', 'Q2', 'Q3', 'Q4', 'Q1 r', 'Q2 r', 'Q3 r', 'Q4 r']:
        quarter_num = str(q_val).strip().replace(' r', '').replace('Q', '')
        year_clean = str(y_val).strip().replace(' r', '')
        records.append({'quarter': f"{year_clean}Q{quarter_num}", 'hk_gdp': gdp_row[col]})

hk_gdp = pd.DataFrame(records)
hk_gdp['hk_gdp'] = pd.to_numeric(hk_gdp['hk_gdp'], errors='coerce')
hk_gdp = hk_gdp.sort_values('quarter').reset_index(drop=True)

print(hk_gdp.shape)
print(hk_gdp.head())
print(hk_gdp.tail())
```

```python
hk_gdp.to_csv('Data/Raw/hongkong_gdp.csv', index=False)
```

```python
import urllib.request
import json
import pandas as pd

url = "https://api.hkma.gov.hk/public/market-data-and-statistics/monthly-statistical-bulletin/er-ir/hkd-ir-effdates?offset=0"

req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
with urllib.request.urlopen(req) as response:
    data = json.loads(response.read().decode())

print(data['result'].keys())
print(data['result']['records'][:3])
```

```python
df = pd.DataFrame(data['result']['records'])
df = df[['effect_date', 'best_lending_rate']].dropna(subset=['best_lending_rate'])
df['effect_date'] = pd.to_datetime(df['effect_date'])
df = df.sort_values('effect_date').reset_index(drop=True)

print(df.shape)
print(df.head())
print(df.tail())
```

```python
# Build a full quarterly range spanning your data
quarters = pd.period_range(df['effect_date'].min(), '2026Q3', freq='Q')
quarterly_df = pd.DataFrame({'quarter': quarters.astype(str)})

# Convert effect_date to the quarter it falls in
df['quarter'] = df['effect_date'].dt.to_period('Q').astype(str)

# For each quarter, take the rate that was in effect at the end of that quarter
quarter_end_rates = df.groupby('quarter')['best_lending_rate'].last().reset_index()
quarter_end_rates = quarter_end_rates.rename(columns={'best_lending_rate': 'hk_lending_rate'})

# Merge onto the full quarter range and forward-fill any quarters with no rate change
hk_lending_rate = quarterly_df.merge(quarter_end_rates, on='quarter', how='left')
hk_lending_rate['hk_lending_rate'] = hk_lending_rate['hk_lending_rate'].ffill()

print(hk_lending_rate.head())
print(hk_lending_rate.tail())
print(hk_lending_rate.isnull().sum())
```

```python
hk_lending_rate.to_csv('Data/Raw/hongkong_lending_rate.csv', index=False)
```

```python
import pandas as pd

hk_price_index = pd.read_csv('Data/Raw/hongkong_price_index.csv')
hk_cpi = pd.read_csv('Data/Raw/hongkong_cpi.csv')
hk_gdp = pd.read_csv('Data/Raw/hongkong_gdp.csv')
hk_income = pd.read_csv('Data/Raw/hongkong_income.csv')
hk_lending_rate = pd.read_csv('Data/Raw/hongkong_lending_rate.csv')

# Merge on 'quarter', inner join keeps only overlapping quarters across all five
hk_merged = hk_price_index.merge(hk_cpi, on='quarter', how='inner')
hk_merged = hk_merged.merge(hk_gdp, on='quarter', how='inner')
hk_merged = hk_merged.merge(hk_income, on='quarter', how='inner')
hk_merged = hk_merged.merge(hk_lending_rate, on='quarter', how='inner')

print(hk_merged.shape)
print(hk_merged['quarter'].min(), '-', hk_merged['quarter'].max())
hk_merged.head()
hk_merged.tail()
```

```python
print(hk_merged.dtypes)
hk_merged.describe()

# Check for duplicates
print(hk_merged['quarter'].duplicated().sum())

# Check for gaps in the quarter sequence
expected = pd.period_range('2015Q1', '2026Q2', freq='Q').astype(str)
actual = set(hk_merged['quarter'])
missing_quarters = [q for q in expected.tolist() if q not in actual]
print(missing_quarters)

# Check for nulls
print(hk_merged.isnull().sum())
```
